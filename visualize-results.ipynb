# Visualize Results of 4-Tank Simulation

```python
from pathlib import Path

import casadi as cas
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from feed_conc_ctrl.plot_utils import make_tsplots
```

```python
sim_name = "do_mpc_demo_4_tanks"
results_dir = Path("results") / sim_name
plot_dir = results_dir / "plots"
plot_dir.mkdir(exist_ok=True, parents=True)
```

```python
filename = "sim_results.csv"
sim_results = pd.read_csv(results_dir / filename, header=[0, 1], index_col=0)
sim_results.info()
```

## Plot Simulation Results

```python
# Units for plot axes labels
units = {
    ("states", "tank_1_v_dot_in"): "$m^3/hr$",
    ("inputstank_2_v_dot_in"): "$m^3/hr$",
    ("inputstank_3_v_dot_in"): "$m^3/hr$",
    ("inputs", "mixer_v_dot_in_1"): "$m^3/hr$",
    ("inputs", "mixer_v_dot_in_2"): "$m^3/hr$",
    ("inputs", "tank_4_v_dot_out"): "$m^3/hr$",
    ("true_outputs", "mixer_v_dot_out"): "$m^3/hr$",
    ("states", "tank_1_L"): "$m$",
    ("states", "tank_2_L"): "$m$",
    ("states", "tank_3_L"): "$m$",
    ("states", "tank_4_L"): "$m$",
    ("states", "tank_1_conc_in"): "$kg/m^3$",
    ("true_outputs", "tank_1_conc_out"): "$kg/m^3$",
    ("true_outputs", "tank_2_conc_out"): "$kg/m^3$",
    ("true_outputs", "tank_3_conc_out"): "$kg/m^3$",
    ("true_outputs", "tank_4_conc_out"): "$kg/m^3$",
    ("time_varying_params", "tank_4_v_dot_out_sp"): "$m^3/hr$",
    ("time_varying_params", "tank_4_conc_out_sp"): "$kg/m^3$",
}
```

```python
plot_info = {
    "Tank 1 Feed Rate": {
        "Flow rate": {
            "var_name": ("states", "tank_1_v_dot_in"),
            "kind": "step",
        },
    },
    "Tank 1 Feed Concentration": {
        "Concentration": {
            "var_name": ("states", "tank_1_conc_in"),
            "kind": "step",
        },
    },
}
make_tsplots(
    sim_results,
    plot_info,
    units=units,
    time_units="hours",
    legend_loc="outside right",
    alpha=0.5,
)
plt.tight_layout()
filename = "tank_1_feed_rate_conc.png"
plt.savefig(plot_dir / filename, dpi=150)
plt.show()
```

```python
plot_info = {
    "Tank Transfer Flow Rates": {
        "Tank 1 to 2": {"var_name": ("inputs", "tank_2_v_dot_in")},
        "Tank 1 to 3": {"var_name": ("inputs", "tank_3_v_dot_in")},
        "Tank 2 to 4": {"var_name": ("inputs", "mixer_v_dot_in_1")},
        "Tank 3 to 4": {"var_name": ("inputs", "mixer_v_dot_in_2")},
    },
    "Tank Levels": {
        "Tank 1": {"var_name": ("true_outputs", "tank_1_L")},
        "Tank 2": {"var_name": ("true_outputs", "tank_2_L")},
        "Tank 3": {"var_name": ("true_outputs", "tank_3_L")},
        "Tank 4": {"var_name": ("true_outputs", "tank_4_L")},
    },
    "Tank Concentrations": {
        "Tank 1": {"var_name": ("true_outputs", "tank_1_conc_out")},
        "Tank 2": {"var_name": ("true_outputs", "tank_2_conc_out")},
        "Tank 3": {"var_name": ("true_outputs", "tank_3_conc_out")},
        "Tank 4": {"var_name": ("true_outputs", "tank_4_conc_out")},
    },
}
make_tsplots(
    sim_results,
    plot_info,
    units=units,
    time_units="hours",
    legend_loc="outside right",
    alpha=0.5,
)
plt.tight_layout()
filename = "tank_flows_levels_concs.png"
plt.savefig(plot_dir / filename, dpi=150)
plt.show()
```

```python
plot_info = {
    "Tank 4 Outflow Rate": {
        #        "In": {'var_name': ('states', 'tank_1_v_dot_in'), 'kind': 'step'},
        #        "In": {'var_name': ('states', 'tank_1_conc_in'), 'kind': 'step'},
        "Actual": {"var_name": ("inputs", "tank_4_v_dot_out")},
        "Setpoint": {
            "var_name": ("time_varying_params", "tank_4_v_dot_out_sp"),
            "style": "k--",
        },
    },
    "Tank 4 Outflow Concentration": {
        "Actual": {"var_name": ("true_outputs", "tank_4_conc_out")},
        "Setpoint": {
            "var_name": ("time_varying_params", "tank_4_conc_out_sp"),
            "style": "k--",
        },
    },
}
make_tsplots(
    sim_results,
    plot_info,
    units=units,
    time_units="hours",
    legend_loc="outside right",
    alpha=0.5,
)
plt.tight_layout()
filename = "tank_4_out_rate_conc.png"
plt.savefig(plot_dir / filename, dpi=150)
plt.show()
```
