# Visualize Results of 4-Tank Simulation

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import casadi as cas
import casadi.tools as castools
from feed_conc_ctrl.plot_utils import make_tsplots
```

```python
sim_name = "do_mpc_demo_4_tanks"
results_dir = Path("results") / sim_name
plot_dir = results_dir / "plots"
plot_dir.mkdir(exist_ok=True, parents=True)
```

```python
filename = "sim_results.csv"
sim_results = pd.read_csv(results_dir / filename, header=[0, 1], index_col=0)
sim_results.info()
```

```python
plot_info = {
    "Flow Rates": {
        "Tank 1 feed": {'var_name': ('states', 'tank_1_v_dot_in'), 'kind': 'step'},
        "Tank 2 feed": {'var_name': ('inputs', 'tank_2_v_dot_in')},
        "Tank 3 feed": {'var_name': ('inputs', 'tank_3_v_dot_in')},
        #"Tank 4 feed": {'var_name': ('mixer_v_dot_out')},
        "Tank 4 out": {'var_name': ('inputs', 'tank_4_v_dot_out')}
    },
    "Tank Levels": {
        "Tank 1": {'var_name': ('true_outputs', 'tank_1_L')},
        "Tank 2": {'var_name': ('true_outputs', 'tank_2_L')},
        "Tank 3": {'var_name': ('true_outputs', 'tank_3_L')},
        "Tank 4": {'var_name': ('true_outputs', 'tank_4_L')}
    },
    "Concentrations": {
        "Tank 1 feed": {'var_name': ('states', 'tank_1_conc_in'), 'kind': 'step'},
        "Tank 1": {'var_name': ('true_outputs', 'tank_1_conc_out')},
        "Tank 2": {'var_name': ('true_outputs', 'tank_2_conc_out')},
        "Tank 3": {'var_name': ('true_outputs', 'tank_3_conc_out')},
        "Tank 4": {'var_name': ('true_outputs', 'tank_4_conc_out')}
    }
}
units = {
    'tank_1_v_dot_in': '$m^3/hr$',
    'tank_2_v_dot_in': '$m^3/hr$',
    'tank_3_v_dot_in': '$m^3/hr$',
    'tank_4_v_dot_in': '$m^3/hr$',
    'mixer_v_dot_in_1': '$m^3/hr$',
    'mixer_v_dot_in_2': '$m^3/hr$',
    'mixer_v_dot_out': '$m^3/hr$',
    'tank_1_L': '$m$',
    'tank_2_L': '$m$',
    'tank_3_L': '$m$',
    'tank_4_L': '$m$',
    'tank_1_conc_in': '$kg/m^3$',
    'tank_1_conc_out': '$kg/m^3$',
    'tank_2_conc_out': '$kg/m^3$',
    'tank_3_conc_out': '$kg/m^3$',
    'tank_4_conc_out': '$kg/m^3$'
}
make_tsplots(
    sim_results,
    plot_info,
    units=units,
    time_units="hours"
)
plt.tight_layout()
plt.show()
```
